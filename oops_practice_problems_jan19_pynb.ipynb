```python
class Animal:
    def __init__(self, name, species, sound):
        self.name = name
        self.species = species
        self.sound = sound

    def speak(self):
        return f"{self.name} {self.species} makes a {self.sound} sound"

jimmy = Animal("jimmy", "dog", "woof")
jimmy.speak()
```

```python
class Student:
  def __init__(self, name, age, gpa):
    self.name = name
    self.age = age
    self.gpa = gpa

  def check_eligibility(self):
    if self.gpa > 9.0:
      print(f"{self.name} is eligible for scholarship")
    else:
      print(f"{self.name} is not eligible for scholarsehip")

  student1 = Student("alvin", 20, 9.2)
  student1.check_eligibility()

  student2 = Student("alvin", 20, 8.5)
  student2.check_eligibility()
```

```python
class Employee:
    def __init__(self, name, experience):
        self.name = name
        self.experience = experience

    def find_senior(self, other):
        if self.experience > other.experience:
            print(f"{self.name} is senior to {other.name}")
        elif self.experience < other.experience:
            print(f"{other.name} is senior to {self.name}")
        else:
            print(f"{self.name} and {other.name} have the same experience")



emp1 = Employee("anna", 8)
emp2 = Employee("john", 5)
emp1.find_senior(emp2)

emp3 = Employee("Mike", 5)
emp2. find_senior(emp3)
```
