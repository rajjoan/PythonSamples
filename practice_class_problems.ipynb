```python
class Student:
    def __init__(self, name, grade, gpa):
        self.name = name
        self.grade = grade
        self.gpa = gpa
    def who_is_senior(self, other_student):
        if self.grade > other_student.grade:
            return self.name
        elif self.grade < other_student.grade:
            return other_student.name
        else:
            return "Both students are in the same grade"
    def print_gpa_scale(self):
      if self.gpa >= 3.5:
        print("perfect")
      elif self.gpa >= 3.0:
        print("good")
      elif self.gpa >= 2:
        print("average")
      else:
        print("bad")


student1= Student("john", 12, 3.8)
student2 = Student("roy", 11, 3.2)
print("senior student is", student1.who_is_senior(student2))
student1.print_gpa_scale()
```
